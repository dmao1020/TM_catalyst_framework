```python
# Ureate ligand SMILES, bidentate
ureate_ligand1_SMILES = "C/[N]([*])=C(N([*])(C)C)/O[*]"#"CC(C1=CC=CC=C1)N(/C(N(C2=C(C=CC=C2C)C)[*])=O\[*])C"
Ligand(ureate_ligand1_SMILES).draw_ligand()
```

```python
# Ureate ligand SMILES, with more than two dummy atoms
ureate_ligand1_SMILES ="C/[N]([*])=C(N([*])(C)C)/O[*]"#"CC(C1=CC=CC=C1)N[*](/C(N(C2=C(C=CC=C2C(C)C)C(C)C)[*])=O\[*])C"
Ligand(ureate_ligand1_SMILES).draw_ligand()
```

```python
# prepare array of ligands with various substituents
# remember to assign proper charge on over valence

ureate_core_smiles = "[*:1]/[N+]([*])=C([N+]([*])([*:2])[*:3])/O[*]"
ureate_core_mol = Chem.MolFromSmiles(ureate_core_smiles, sanitize = False) 

ureate_core_mol
```

```python
R1_smiles = "CC(C1=CC=CC=C1)[*:1]"
R2_smiles = "[*:2]C2=C(C=CC=C2C(C)C)C(C)C"
R3_smiles = "C[*:3]"

ligand_util = Ligand()
full_ligand_smiles = ligand_util.attach_R_group(
    core_smiles = ureate_core_smiles,
    R_smiles_ls = [R1_smiles, R2_smiles, R3_smiles]
)
print (f"Ligand SMILES: {full_ligand_smiles}")
ligand_util.ligand
```
