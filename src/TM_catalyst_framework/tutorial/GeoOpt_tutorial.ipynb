```python
from rdkit import Chem
from rdkit.Chem.EnumerateStereoisomers import EnumerateStereoisomers, StereoEnumerationOptions

from TM_catalyst_framework.io_utils import mol_visual
import openbabel.openbabel as ob
from openbabel import pybel
general_ureate_ligand_smiles = "CC([N+](C(O[*])=NC1=C(C(C)C)C=CC=C1C(C)C)(C)[*])C2=CC=CC=C2"
#"CN(CC)C(N(C(C)C))=O"          

m = Chem.MolFromSmiles(general_ureate_ligand_smiles)
isomers = tuple(EnumerateStereoisomers(m))

ureate_iso_idx = 0
ligand_smiles = sorted(Chem.MolToSmiles(x, isomericSmiles=True) for x in isomers)[ureate_iso_idx]

# ligand_smiles_clean = ligand_smiles.replace("+", "")

ligand = pybel.readstring("smi", ligand_smiles)
```

```python

ligand.make3D()
ligand_3d = mol_visual.visualize_mol_molecule(ligand)
ligand_3d
```

```python
print (f"Optimizing ligand geometry with metal constraints for ligand: {ligand_smiles}")
GeoOpt_util = GeoOpt(force_field = "uff",
                    force_field_steps = 2000,
                    metal_center = "Ti")

opt_ligand_mol = GeoOpt_util.organic_opt(smiles = ligand_smiles)
opt_ligand_3d = mol_visual.visualize_mol_molecule(opt_ligand_mol)

opt_ligand_3d
```

```python
ff = ob.OBForceField.FindForceField("uff")
success = ff.Setup(ligand.OBMol)
if success:
    print (success)
    print(ff.Energy())
    ff.ConjugateGradients(5000)
    print(ff.Energy())
    ff.GetCoordinates(ligand.OBMol)
else:
    print ("Failed to set up force field")
# ligand_3d_opt = mol_visual.visualize_mol_molecule(ligand)
# ligand_3d_opt
```
